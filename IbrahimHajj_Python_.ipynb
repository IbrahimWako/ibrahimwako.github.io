```python
"""
Receipt Calculator
This program makes a receipt for 3 customers in a stationery shop.
It finds the line totals, subtotal, discount, VAT and total to pay.
Written by: Ibrahim
"""

# Global variables
shop_name = "Campus Stationery Shop"
vat_rate = 0.075

# Prices of the five items (NGN)
notebook_price = 1500
pen_pack_price = 800
usb_price = 6500
calculator_price = 9000
backpack_price = 22000

# Discount rates for the three customers (multiple assignment)
discount1, discount2, discount3 = 0.10, 0, 0.10

# ----- Customer 1 -----
# Name and quantities
name1 = "Amaka"
notebook_qty1, pen_pack_qty1, usb_qty1, calculator_qty1, backpack_qty1 = 3, 2, 0, 1, 0

# Line totals (price x quantity)
notebook_total1 = notebook_price * notebook_qty1
pen_pack_total1 = pen_pack_price * pen_pack_qty1
usb_total1 = usb_price * usb_qty1
calculator_total1 = calculator_price * calculator_qty1
backpack_total1 = backpack_price * backpack_qty1

# Subtotal, discount, VAT and total
subtotal1 = notebook_total1 + pen_pack_total1 + usb_total1 + calculator_total1 + backpack_total1
discount_amount1 = subtotal1 * discount1
after_discount1 = subtotal1 - discount_amount1
vat_amount1 = after_discount1 * vat_rate
total_to_pay1 = after_discount1 + vat_amount1

# Print receipt 1
print("=====", shop_name, "=====")
print("Customer:", name1)
print("Notebook:", notebook_qty1, "x", notebook_price, "=", notebook_total1)
print("Pen pack:", pen_pack_qty1, "x", pen_pack_price, "=", pen_pack_total1)
print("USB flash drive:", usb_qty1, "x", usb_price, "=", usb_total1)
print("Calculator:", calculator_qty1, "x", calculator_price, "=", calculator_total1)
print("Backpack:", backpack_qty1, "x", backpack_price, "=", backpack_total1)
print("Subtotal:", subtotal1)
print("Discount:", round(discount_amount1, 2))
print("After discount:", round(after_discount1, 2))
print("VAT:", round(vat_amount1, 2))
print("Total to pay:", round(total_to_pay1, 2))
print()

# ----- Customer 2 -----
# Name and quantities
name2 = "Bola"
notebook_qty2, pen_pack_qty2, usb_qty2, calculator_qty2, backpack_qty2 = 0, 0, 1, 0, 1

# Line totals (price x quantity)
notebook_total2 = notebook_price * notebook_qty2
pen_pack_total2 = pen_pack_price * pen_pack_qty2
usb_total2 = usb_price * usb_qty2
calculator_total2 = calculator_price * calculator_qty2
backpack_total2 = backpack_price * backpack_qty2

# Subtotal, discount, VAT and total
subtotal2 = notebook_total2 + pen_pack_total2 + usb_total2 + calculator_total2 + backpack_total2
discount_amount2 = subtotal2 * discount2
after_discount2 = subtotal2 - discount_amount2
vat_amount2 = after_discount2 * vat_rate
total_to_pay2 = after_discount2 + vat_amount2

# Print receipt 2
print("=====", shop_name, "=====")
print("Customer:", name2)
print("Notebook:", notebook_qty2, "x", notebook_price, "=", notebook_total2)
print("Pen pack:", pen_pack_qty2, "x", pen_pack_price, "=", pen_pack_total2)
print("USB flash drive:", usb_qty2, "x", usb_price, "=", usb_total2)
print("Calculator:", calculator_qty2, "x", calculator_price, "=", calculator_total2)
print("Backpack:", backpack_qty2, "x", backpack_price, "=", backpack_total2)
print("Subtotal:", subtotal2)
print("Discount:", round(discount_amount2, 2))
print("After discount:", round(after_discount2, 2))
print("VAT:", round(vat_amount2, 2))
print("Total to pay:", round(total_to_pay2, 2))
print()

# ----- Customer 3 -----
# Name and quantities
name3 = "Chidi"
notebook_qty3, pen_pack_qty3, usb_qty3, calculator_qty3, backpack_qty3 = 10, 5, 2, 0, 1

# Line totals (price x quantity)
notebook_total3 = notebook_price * notebook_qty3
pen_pack_total3 = pen_pack_price * pen_pack_qty3
usb_total3 = usb_price * usb_qty3
calculator_total3 = calculator_price * calculator_qty3
backpack_total3 = backpack_price * backpack_qty3

# Subtotal, discount, VAT and total
subtotal3 = notebook_total3 + pen_pack_total3 + usb_total3 + calculator_total3 + backpack_total3
discount_amount3 = subtotal3 * discount3
after_discount3 = subtotal3 - discount_amount3
vat_amount3 = after_discount3 * vat_rate
total_to_pay3 = after_discount3 + vat_amount3

# Print receipt 3
print("=====", shop_name, "=====")
print("Customer:", name3)
print("Notebook:", notebook_qty3, "x", notebook_price, "=", notebook_total3)
print("Pen pack:", pen_pack_qty3, "x", pen_pack_price, "=", pen_pack_total3)
print("USB flash drive:", usb_qty3, "x", usb_price, "=", usb_total3)
print("Calculator:", calculator_qty3, "x", calculator_price, "=", calculator_total3)
print("Backpack:", backpack_qty3, "x", backpack_price, "=", backpack_total3)
print("Subtotal:", subtotal3)
print("Discount:", round(discount_amount3, 2))
print("After discount:", round(after_discount3, 2))
print("VAT:", round(vat_amount3, 2))
print("Total to pay:", round(total_to_pay3, 2))
```

**What is a variable?**

A variable is a name I use to keep a value in the computer. It is like a box with a label on it. When I write notebook_price = 1500, the number 1500 goes into the box called notebook_price. Later I can just use the name, and I can also put a new value inside.

**What are the rules for naming a variable?**

1. The name must start with a letter or an underscore (_). It cannot start with a number.

2. The name can only have letters, numbers and underscores. No spaces and no signs like - or $.

3. Small and capital letters are different. price and Price are two different names.

4. I cannot use words that Python already owns, like if, for or class.

5. The name should tell me what it is for, like total_to_pay, not just x.

**What is a global variable and what is a local variable?** 

A global variable is made outside a function, so the whole program can use it. For example, the name of the shop on the signboard. Everybody can see it.

A local variable is made inside a function, so only that function can use it. For example, a small paper a cashier uses to add one customer's bill. When the bill is finished, the paper is thrown away.

If I use a local variable outside the function where I made it, Python shows a NameError. This is because the name does not exist outside the function.

**Why should you never use a reserved name, such as `int` or `float`, as a variable name?**

Python already uses names like int and float to do special work. For example, int changes a value into a whole number. If I use int as my own variable name, Python forgets its special work and only sees my value. For example, if I write int = 5, then int("10") will not work anymore and I will get an error. Some words, like if and for, are not allowed at all, and Python gives a SyntaxError at once. So it is better to use my own clear names.
